```python
import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.monitor import Monitor

# 8 環境を並列に走らせる（方策オン型はデータ収集が律速なので効果が大きい）
vec_env = make_vec_env("CartPole-v1", n_envs=8, seed=42)

model = PPO(
    "MlpPolicy", vec_env,
    learning_rate=3e-4,
    n_steps=256,        # 各環境で何ステップ集めてから更新するか
    batch_size=256,
    n_epochs=10,        # 集めたデータを何周学習するか（PPO は再利用できる）
    gamma=0.99,
    gae_lambda=0.95,    # GAE：バイアスと分散のトレードオフ（1 に近いほど低バイアス高分散）
    clip_range=0.2,     # ← PPO のクリップ幅 ε
    ent_coef=0.0,       # エントロピーボーナス。探索が足りないときは 0.01 程度に
    verbose=1, seed=42,
)
model.learn(total_timesteps=100_000)
```

```python
# --- 評価は必ず別環境・決定的方策で ---
eval_env = Monitor(gym.make("CartPole-v1"))
eval_env.reset(seed=1000)
mean_r, std_r = evaluate_policy(model, eval_env, n_eval_episodes=20,
                                deterministic=True)
print(f"平均得点: {mean_r:.1f} ± {std_r:.1f}")     # 500 に近ければ成功

model.save("ppo_cartpole")
eval_env.close()
vec_env.close()
```

### ランダム方策との比較 — ベースラインを必ず並べる

```python
import numpy as np

env = gym.make("CartPole-v1")
env.action_space.seed(1000)
rs = []
for i in range(20):
    obs, _ = env.reset(seed=1000 + i)
    done, total = False, 0.0
    while not done:
        obs, r, term, trunc, _ = env.step(env.action_space.sample())
        total += r; done = term or trunc
    rs.append(total)
env.close()

print(f"ランダム方策: {np.mean(rs):6.1f} ± {np.std(rs):.1f}")
print(f"PPO         : {mean_r:6.1f} ± {std_r:.1f}")
print(f"→ {mean_r / np.mean(rs):.1f} 倍")
```

## 読み取りどころ

- **`n_envs=8` の並列化が効くのは、PPO が方策オン型だから。**いま学習している方策で集めたデータしか使えないので、**データ収集が律速**になる。DQN のような方策オフ型では、過去のデータを再利用できるのでここまでの差は出ない。
- **`n_epochs=10` は「集めたデータを 10 周使う」という意味。**方策オン型でありながら再利用できるのが PPO の工夫で、**`clip_range` がその再利用による方策の暴走を抑えている**。
- ⚠️ **評価は `deterministic=True` かつ別環境で。**学習中の環境は探索でノイズが乗っているし、確率的方策のままだと得点がぶれる。
- ⚠️ **`verbose=1` のログに出る `ep_rew_mean` は学習中の値**であって、評価値ではない。報告に使うなら `evaluate_policy` の結果のほうである。

⚠️ 強化学習は seed でぶれる。**同じ設定で 3〜5 回走らせて中央値を見る**のが実務の作法。
