```python
import gymnasium as gym
from stable_baselines3 import SAC

# Pendulum は連続行動（トルクの実数値）— DQN では扱えない課題
env = gym.make("Pendulum-v1")
print("行動空間:", env.action_space, "← Box = 連続値。Discrete ではない")
print("観測空間:", env.observation_space)
env.close()
```

```python
env = gym.make("Pendulum-v1")
model = SAC("MlpPolicy", env, learning_rate=3e-4, buffer_size=100_000,
            batch_size=256, gamma=0.99, tau=0.005, verbose=1, seed=42)
model.learn(total_timesteps=30_000)
```

```python
import numpy as np
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.monitor import Monitor

eval_env = Monitor(gym.make("Pendulum-v1"))
eval_env.reset(seed=500)
eval_env.action_space.seed(500)
mean_r, std_r = evaluate_policy(model, eval_env, n_eval_episodes=10,
                                deterministic=True)

# ベースライン：ランダム方策
rs = []
for i in range(10):
    obs, _ = eval_env.reset(seed=500 + i)
    done, total = False, 0.0
    while not done:
        obs, r, term, trunc, _ = eval_env.step(eval_env.action_space.sample())
        total += r; done = term or trunc
    rs.append(total)
eval_env.close()

print(f"ランダム方策: {np.mean(rs):8.1f} ± {np.std(rs):.1f}")
print(f"SAC         : {mean_r:8.1f} ± {std_r:.1f}")
print("\n※ Pendulum の報酬は負の値。0 に近いほど良い（棒が立っている）")

model.save("sac_pendulum")
env.close()
```

## 読み取りどころ

- **`Box` な行動空間は DQN では扱えない。**DQN は「全行動の Q 値を出して argmax」という設計なので、行動が無限にあると成立しない。**連続行動には SAC / TD3 / PPO を使う。**
- **`buffer_size` は方策オフ型だから意味を持つ。**過去の経験を再利用できるので、サンプル効率が PPO より高い。**実機やシミュレータが高価な場面では SAC 系が選ばれる。**
- **`tau=0.005` はターゲットネットワークのソフト更新率。**DQN が「定期的に丸ごとコピー」だったのに対し、SAC は毎ステップ少しずつ寄せる。
- ⚠️ **Pendulum の報酬は負の値。**「大きいほど良い」は変わらないが、0 に近づくのがゴールなので、**グラフの見え方が CartPole と逆**になる。

⚠️ SAC は**エントロピー係数を自動調整する**（既定 `ent_coef="auto"`）。探索の強さを自分で調整しなくてよいのが利点だが、**うまくいかないときに何が原因か分かりにくい**という裏返しもある。
